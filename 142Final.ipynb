```python
import os, numpy as np, torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

ROOT = "/content/drive/MyDrive/GeoSR-AI"
DATA = ROOT + "/data/training_x4"
SAVE = ROOT + "/models/GeoSRNet_x4_best.pth"

device = "cuda"
torch.backends.cudnn.benchmark = True

# Dataset
class SRDataset(Dataset):
    def __init__(self, split):
        self.lr = np.load(f"{DATA}/LR_{split}.npy", mmap_mode="r")
        self.hr = np.load(f"{DATA}/HR_{split}.npy", mmap_mode="r")

    def __len__(self):
        return len(self.hr)

    def __getitem__(self, i):
        lr = torch.from_numpy(np.array(self.lr[i], copy=True)).permute(2,0,1)
        hr = torch.from_numpy(np.array(self.hr[i], copy=True)).permute(2,0,1)

        if torch.rand(()) > 0.5:
            lr, hr = lr.flip(-1), hr.flip(-1)
        if torch.rand(()) > 0.5:
            lr, hr = lr.flip(-2), hr.flip(-2)

        return lr.float(), hr.float()

train_loader = DataLoader(SRDataset("train"), batch_size=4, shuffle=True, num_workers=2, pin_memory=True)
val_loader = DataLoader(SRDataset("val"), batch_size=4, shuffle=False, num_workers=2, pin_memory=True)

# EDSR-lite ×4
class ResBlock(nn.Module):
    def __init__(self, c):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(c,c,3,padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(c,c,3,padding=1)
        )

    def forward(self,x):
        return x + 0.1*self.net(x)

class GeoSRNet(nn.Module):
    def __init__(self):
        super().__init__()
        c = 48
        self.head = nn.Conv2d(4,c,3,padding=1)
        self.body = nn.Sequential(*[ResBlock(c) for _ in range(8)])
        self.up = nn.Sequential(
            nn.Conv2d(c,c*4,3,padding=1),
            nn.PixelShuffle(2),
            nn.ReLU(inplace=True),
            nn.Conv2d(c,c*4,3,padding=1),
            nn.PixelShuffle(2),
            nn.ReLU(inplace=True)
        )
        self.tail = nn.Conv2d(c,4,3,padding=1)

    def forward(self,x):
        base = F.interpolate(x, scale_factor=4, mode="bicubic", align_corners=False)
        f = self.head(x)
        f = f + self.body(f)
        return base + self.tail(self.up(f))

model = GeoSRNet().to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, patience=4, factor=0.5
)
scaler = torch.amp.GradScaler("cuda")

best = float("inf")
EPOCHS = 30

for epoch in range(EPOCHS):
    model.train()
    train_loss = 0

    for lr, hr in train_loader:
        lr, hr = lr.to(device), hr.to(device)
        optimizer.zero_grad(set_to_none=True)

        with torch.autocast("cuda", dtype=torch.float16):
            pred = model(lr)
            loss = F.l1_loss(pred, hr)

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        scaler.step(optimizer)
        scaler.update()
        train_loss += loss.item()

    model.eval()
    val_loss = 0

    with torch.no_grad():
        for lr, hr in val_loader:
            lr, hr = lr.to(device), hr.to(device)
            with torch.autocast("cuda", dtype=torch.float16):
                pred = model(lr)
                val_loss += F.l1_loss(pred, hr).item()

    train_loss /= len(train_loader)
    val_loss /= len(val_loader)
    scheduler.step(val_loss)

    if val_loss < best:
        best = val_loss
        torch.save(model.state_dict(), SAVE)

    print(f"Epoch {epoch+1}/{EPOCHS} | Train: {train_loss:.6f} | Val: {val_loss:.6f} | Best: {best:.6f}")

print("Training complete.")
print("Best model:", SAVE)
```

```python
import numpy as np, torch
import torch.nn.functional as F
from skimage.metrics import peak_signal_noise_ratio, structural_similarity

model = GeoSRNet().cuda()
model.load_state_dict(torch.load(
    "/content/drive/MyDrive/GeoSR-AI/models/GeoSRNet_x4_best.pth",
    map_location="cuda"
))
model.eval()

lr = torch.tensor(np.load(
    "/content/drive/MyDrive/GeoSR-AI/data/training_x4/LR_val.npy"
)).permute(0,3,1,2).cuda()

hr = np.load("/content/drive/MyDrive/GeoSR-AI/data/training_x4/HR_val.npy")

with torch.no_grad():
    pred = model(lr).clamp(0,1).cpu().permute(0,2,3,1).numpy()
    base = F.interpolate(lr, scale_factor=4, mode="bicubic",
                         align_corners=False).clamp(0,1)
    base = base.cpu().permute(0,2,3,1).numpy()

for name, result in [("Bicubic",base),("GeoSRNet ×4",pred)]:
    psnr, ssim = [], []
    for i in range(len(hr)):
        psnr.append(peak_signal_noise_ratio(hr[i],result[i],data_range=1))
        ssim.append(structural_similarity(hr[i],result[i],
                     channel_axis=-1,data_range=1))
    print(name, "PSNR:",round(np.mean(psnr),3),
          "SSIM:",round(np.mean(ssim),4))
```

```python
import numpy as np

eps = 1e-8

def sam(a,b):
    dot = np.sum(a*b,axis=-1)
    na = np.linalg.norm(a,axis=-1)
    nb = np.linalg.norm(b,axis=-1)
    return np.degrees(np.arccos(np.clip(dot/(na*nb+eps),-1,1)))

def ndvi(x):
    return (x[...,3]-x[...,2])/(x[...,3]+x[...,2]+eps)

print("SAM:", np.mean(sam(hr,pred)))
print("NDVI RMSE:", np.sqrt(np.mean((ndvi(hr)-ndvi(pred))**2)))
```

```python
import os, numpy as np, torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset

# 1. Load synthetic ×4 dataset
root = "/content/drive/MyDrive/GeoSR-AI/data/training_x4"

def load_data(name):
    a = np.load(f"{root}/{name}.npy")
    return torch.from_numpy(a).float().permute(0,3,1,2).contiguous()

HR_train, LR_train = load_data("HR_train"), load_data("LR_train")
HR_val, LR_val = load_data("HR_val"), load_data("LR_val")

print("Training:", HR_train.shape, LR_train.shape)
print("Validation:", HR_val.shape, LR_val.shape)

# 2. Restore original model
class ResBlock(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(48,48,3,padding=1), nn.ReLU(inplace=True),
            nn.Conv2d(48,48,3,padding=1)
        )
    def forward(self,x):
        return x + 0.1*self.net(x)

class GeoSRNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.head = nn.Conv2d(4,48,3,padding=1)
        self.body = nn.Sequential(*[ResBlock() for _ in range(8)])
        self.up = nn.Sequential(
            nn.Conv2d(48,192,3,padding=1), nn.PixelShuffle(2), nn.ReLU(),
            nn.Conv2d(48,192,3,padding=1), nn.PixelShuffle(2), nn.ReLU()
        )
        self.tail = nn.Conv2d(48,4,3,padding=1)

    def forward(self,x):
        f = self.head(x)
        f = f + self.body(f)
        return F.interpolate(x,scale_factor=4,mode="bicubic",
                             align_corners=False) + self.tail(self.up(f))

device = "cuda" if torch.cuda.is_available() else "cpu"
model = GeoSRNet().to(device)

original = torch.load(
    "/content/drive/MyDrive/GeoSR-AI/models/GeoSRNet_x4_best.pth",
    map_location="cpu", weights_only=True
)
model.load_state_dict(original)

# 3. Fine-tune with supervised + consistency loss
loader = DataLoader(
    TensorDataset(LR_train, HR_train),
    batch_size=4, shuffle=True
)

optimizer = torch.optim.Adam(model.parameters(), lr=1e-5)
epochs = 5

for epoch in range(epochs):
    model.train()
    total = 0

    for lr, hr in loader:
        lr, hr = lr.to(device), hr.to(device)

        optimizer.zero_grad()
        pred = model(lr)

        supervised = F.l1_loss(pred, hr)
        down = F.interpolate(
            pred, size=lr.shape[-2:],
            mode="bicubic", align_corners=False
        )
        consistency = F.l1_loss(down, lr)

        loss = supervised + consistency
        loss.backward()
        optimizer.step()
        total += loss.item()

    print(f"Epoch {epoch+1}/{epochs} | Loss: {total/len(loader):.6f}")

# 4. Validate
model.eval()
with torch.no_grad():
    lr, hr = LR_val.to(device), HR_val.to(device)
    pred = model(lr)

    model_down = F.interpolate(
        pred, size=lr.shape[-2:],
        mode="bicubic", align_corners=False
    )
    bicubic = F.interpolate(
        lr, size=pred.shape[-2:],
        mode="bicubic", align_corners=False
    )
    bicubic_down = F.interpolate(
        bicubic, size=lr.shape[-2:],
        mode="bicubic", align_corners=False
    )

    print("\nVALIDATION")
    print("Supervised L1:", F.l1_loss(pred, hr).item())
    print("GeoSR consistency MAE:", F.l1_loss(model_down, lr).item())
    print("Bicubic consistency MAE:", F.l1_loss(bicubic_down, lr).item())

# 5. Save separately
save_path = "/content/drive/MyDrive/GeoSR-AI/models/GeoSRNet_x4_consistency.pth"
torch.save(model.cpu().state_dict(), save_path)

print("\nSaved:", save_path)
```

```python
import rasterio, numpy as np, random
from rasterio.enums import Resampling
from rasterio.windows import Window

random.seed(42)
patches = []
size = 256

with rasterio.open(scl_path) as src:
    scl10 = src.read(
        1,
        out_shape=(10980, 10980),
        resampling=Resampling.nearest
    )

valid = ~np.isin(scl10, [0, 1, 3, 8, 9, 10])

with rasterio.open(paths[0]) as ref:
    H, W = ref.height, ref.width

    for _ in range(500):
        r = random.randrange(0, H-size)
        c = random.randrange(0, W-size)

        mask = valid[r:r+size, c:c+size]
        if mask.mean() < 0.98:
            continue

        bands = []
        for p in paths:
            with rasterio.open(p) as src:
                a = src.read(1, window=Window(c, r, size, size))
            bands.append((a.astype(np.float32)-1000)/20000)

        hr = np.stack(bands)
        hr[:, ~mask] = 0
        patches.append(hr)

        if len(patches) == 16:
            break

print("Accepted patches:", len(patches))
print("Patch array shape:", np.asarray(patches).shape)
print("Valid SCL coverage:", round(float(valid.mean())*100, 2), "%")
if patches:
    print("Reflectance range:", float(np.min(patches)),
          float(np.max(patches)))
```

```python
quality = []

for p in patches:
    valid_pixels = np.broadcast_to(
        valid[0:256, 0:256], p.shape[1:]
    )
    neg_fraction = (p[:, valid_pixels] < -0.01).mean()
    finite = np.isfinite(p).all()

    quality.append({
        "negative_fraction": float(neg_fraction),
        "finite": bool(finite),
        "pass": bool(finite and neg_fraction < 0.005)
    })

print("Total patches:", len(quality))
print("Quality-passed:", sum(x["pass"] for x in quality))
print("Rejected:", sum(not x["pass"] for x in quality))
print("Maximum negative fraction:",
      round(max(x["negative_fraction"] for x in quality)*100, 3), "%")
```

```python
import os, glob, zipfile, shutil, random
import numpy as np
import rasterio, torch
import torch.nn.functional as F
import xml.etree.ElementTree as ET
from rasterio.warp import reproject, Resampling
from rasterio.windows import Window

def build_scene(scene, split, count=8):
    tile = scene.split("_")[-2]

    zp = next(
        p for p in glob.glob(
            "/content/drive/MyDrive/GeoSR-AI/data/sentinel2/**/*.zip",
            recursive=True
        ) if scene in os.path.basename(p)
    )

    tmp = "/content/geosr_current_scene"
    shutil.rmtree(tmp, ignore_errors=True)
    os.makedirs(tmp)

    with zipfile.ZipFile(zp) as z:
        files = z.namelist()
        xml_name = next(n for n in files if n.endswith("MTD_MSIL2A.xml"))

        selected = []
        for b, res in [
            ("B02","10m"), ("B03","10m"),
            ("B04","10m"), ("B08","10m"), ("SCL","20m")
        ]:
            found = [
                n for n in files
                if tile in n and f"_{b}_{res}.jp2" in n
            ]
            assert len(found) == 1, (scene, b, len(found))
            selected.append(found[0])

        for n in selected + [xml_name]:
            z.extract(n, tmp)

        root = ET.fromstring(z.read(xml_name))

    def tag(t):
        return t.split("}")[-1]

    quant = float(next(
        n.text for n in root.iter()
        if tag(n.tag) == "BOA_QUANTIFICATION_VALUE"
    ))

    offset_map = {
        int(n.attrib["band_id"]): float(n.text)
        for n in root.iter()
        if tag(n.tag) == "BOA_ADD_OFFSET"
    }

    offsets = [offset_map[i] for i in [1,2,3,7]]

    paths = [
        glob.glob(tmp + f"/**/*_{b}_10m.jp2", recursive=True)[0]
        for b in ["B02","B03","B04","B08"]
    ]
    scl_path = glob.glob(tmp + "/**/*_SCL_20m.jp2", recursive=True)[0]

    srcs = [rasterio.open(p) for p in paths]
    ref = srcs[0]

    with rasterio.open(scl_path) as ss:
        scl = np.zeros(ref.shape, dtype=np.uint8)
        reproject(
            rasterio.band(ss, 1), scl,
            src_transform=ss.transform, src_crs=ss.crs,
            dst_transform=ref.transform, dst_crs=ref.crs,
            resampling=Resampling.nearest
        )

    valid = ~np.isin(scl, [0,1,3,8,9,10])
    rng = random.Random(sum(map(ord, scene)))

    hrs, masks, coords = [], [], []

    for _ in range(2500):
        r = rng.randrange(0, ref.height-256)
        c = rng.randrange(0, ref.width-256)

        mask = valid[r:r+256, c:c+256].copy()
        if mask.mean() < 0.99:
            continue

        raw = [
            s.read(1, window=Window(c,r,256,256))
            for s in srcs
        ]

        mask &= np.logical_and.reduce([a > 0 for a in raw])
        if mask.mean() < 0.99:
            continue

        hr = np.stack([
            (a.astype(np.float32) + off) / quant / 2
            for a, off in zip(raw, offsets)
        ])

        hr[:, ~mask] = 0
        hrs.append(hr)
        masks.append(mask.astype(np.uint8))
        coords.append([r,c])

        if len(hrs) == count:
            break

    for s in srcs:
        s.close()

    assert len(hrs) >= max(4, count//2), (
        scene, "insufficient clean patches", len(hrs)
    )

    hr = np.asarray(hrs, dtype=np.float32)
    mask = np.asarray(masks, dtype=np.uint8)[:,None]

    lr = F.interpolate(
        torch.from_numpy(hr),
        size=(64,64), mode="bicubic",
        align_corners=False, antialias=True
    ).numpy()

    out = f"/content/drive/MyDrive/GeoSR-AI/data/asia_dataset_v2/{split}"
    os.makedirs(out, exist_ok=True)

    np.savez_compressed(
        os.path.join(out, scene + ".npz"),
        HR=hr, LR=lr, MASK=mask,
        coordinates=np.asarray(coords),
        offsets=np.asarray(offsets),
        quantification=quant
    )

    shutil.rmtree(tmp, ignore_errors=True)

    return len(hr), float(mask.mean()), float(hr.min()), float(hr.max())


result = build_scene(split_data["train_scenes"][0], "train")
print("Patches, valid coverage, minimum, maximum:", result)
```

```python
import torch, numpy as np, glob, os, random
from torch.utils.data import Dataset, DataLoader

class GeoSRDataset(Dataset):
    def __init__(self, folder, augment=False):
        self.files = sorted(glob.glob(folder + "/*.npz"))
        self.index = []
        self.augment = augment
        for i, p in enumerate(self.files):
            with np.load(p) as d:
                self.index.extend((i, j) for j in range(len(d["HR"])))

    def __len__(self):
        return len(self.index)

    def __getitem__(self, idx):
        i, j = self.index[idx]
        with np.load(self.files[i]) as d:
            lr = torch.from_numpy(d["LR"][j].copy())
            hr = torch.from_numpy(d["HR"][j].copy())
            mask = torch.from_numpy(d["MASK"][j].copy()).float()

        if self.augment:
            if random.random() < 0.5:
                lr, hr, mask = lr.flip(-1), hr.flip(-1), mask.flip(-1)
            if random.random() < 0.5:
                lr, hr, mask = lr.flip(-2), hr.flip(-2), mask.flip(-2)
            k = random.randrange(4)
            lr, hr, mask = torch.rot90(lr,k,(-2,-1)), torch.rot90(hr,k,(-2,-1)), torch.rot90(mask,k,(-2,-1))

        return lr.float(), hr.float(), mask

train_ds = GeoSRDataset(f"{backup_root}/train", augment=True)
val_ds = GeoSRDataset(f"{backup_root}/val")

train_loader = DataLoader(train_ds, batch_size=4, shuffle=True, num_workers=0, pin_memory=True)
val_loader = DataLoader(val_ds, batch_size=4, shuffle=False, num_workers=0, pin_memory=True)

lr, hr, mask = next(iter(train_loader))
print("Train patches:", len(train_ds), "| Validation patches:", len(val_ds))
print("Batch:", lr.shape, hr.shape, mask.shape)
print("All finite:", torch.isfinite(lr).all().item(), torch.isfinite(hr).all().item())
```

```python
import sys, random
from torch.utils.data import Dataset, DataLoader

sys.path.insert(0, f"{ROOT}/Final_Model")
from geosrnet import GeoSRNet

class GeoSRDataset(Dataset):
    def __init__(self, folder, augment=False):
        self.files = sorted(glob.glob(folder + "/*.npz"))
        self.index = []
        self.augment = augment

        for i, p in enumerate(self.files):
            with np.load(p) as d:
                self.index.extend((i, j) for j in range(len(d["HR"])))

    def __len__(self):
        return len(self.index)

    def __getitem__(self, idx):
        i, j = self.index[idx]
        with np.load(self.files[i]) as d:
            lr = torch.from_numpy(d["LR"][j].copy()).float()
            hr = torch.from_numpy(d["HR"][j].copy()).float()
            mask = torch.from_numpy(d["MASK"][j].copy()).float()

        if self.augment:
            if random.random() < 0.5:
                lr, hr, mask = lr.flip(-1), hr.flip(-1), mask.flip(-1)
            if random.random() < 0.5:
                lr, hr, mask = lr.flip(-2), hr.flip(-2), mask.flip(-2)
            k = random.randrange(4)
            lr, hr, mask = (
                torch.rot90(x, k, (-2, -1)) for x in (lr, hr, mask)
            )

        return lr, hr, mask

train_ds = GeoSRDataset(f"{backup_root}/train", True)
val_ds = GeoSRDataset(f"{backup_root}/val")

train_loader = DataLoader(
    train_ds, batch_size=4, shuffle=True, num_workers=0, pin_memory=True
)
val_loader = DataLoader(
    val_ds, batch_size=4, shuffle=False, num_workers=0, pin_memory=True
)

model = GeoSRNet().to(device)
checkpoint = f"{ROOT}/Final_Model/GeoSRNet_x4_best.pth"
model.load_state_dict(torch.load(checkpoint, map_location="cpu", weights_only=True))

optimizer = torch.optim.Adam(model.parameters(), lr=2e-5)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode="min", factor=0.5, patience=4
)
scaler = torch.amp.GradScaler("cuda")

print("Model parameters:", sum(p.numel() for p in model.parameters()))
print("Train patches:", len(train_ds))
print("Validation patches:", len(val_ds))
print("Checkpoint loaded successfully")
```

```python
best_val = float("inf")
stale = 0
patience = 10
history = []
epochs = 40

best_path = f"{output_dir}/GeoSRNet_Asia_x4_best.pth"
resume_path = f"{output_dir}/training_resume.pth"

for epoch in range(1, epochs + 1):
    start = time.time()
    model.train()
    train_losses = []

    for lr, hr, mask in train_loader:
        lr, hr, mask = lr.to(device), hr.to(device), mask.to(device)
        optimizer.zero_grad(set_to_none=True)

        with torch.amp.autocast("cuda", dtype=torch.float16):
            pred = model(lr)
            loss = ((pred.float() - hr).abs() * mask).sum() / (mask.sum() * 4)

        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()
        train_losses.append(loss.item())

    model.eval()
    val_losses = []

    with torch.no_grad():
        for lr, hr, mask in val_loader:
            lr, hr, mask = lr.to(device), hr.to(device), mask.to(device)

            with torch.amp.autocast("cuda", dtype=torch.float16):
                pred = model(lr)
                loss = ((pred.float() - hr).abs() * mask).sum() / (mask.sum() * 4)

            val_losses.append(loss.item())

    train_loss = float(np.mean(train_losses))
    val_loss = float(np.mean(val_losses))
    scheduler.step(val_loss)

    improved = val_loss < best_val
    if improved:
        best_val = val_loss
        stale = 0
        torch.save(model.state_dict(), best_path)
    else:
        stale += 1

    history.append({
        "epoch": epoch,
        "train_masked_L1": train_loss,
        "val_masked_L1": val_loss,
        "lr": optimizer.param_groups[0]["lr"]
    })

    with open(f"{output_dir}/training_history.json", "w") as f:
        json.dump(history, f, indent=2)

    torch.save({
        "epoch": epoch,
        "model": model.state_dict(),
        "optimizer": optimizer.state_dict(),
        "scheduler": scheduler.state_dict(),
        "scaler": scaler.state_dict(),
        "best_val": best_val,
        "stale": stale,
        "history": history
    }, resume_path)

    print(
        f"Epoch {epoch:02d}/{epochs} | "
        f"Train {train_loss:.6f} | Val {val_loss:.6f} | "
        f"Best {best_val:.6f} | "
        f"LR {optimizer.param_groups[0]['lr']:.2e} | "
        f"{time.time()-start:.1f}s"
        + (" | SAVED BEST" if improved else "")
    )

    if stale >= patience:
        print("Early stopping triggered.")
        break

print("Training complete. Best validation L1:", best_val)
```

```python
from torch.utils.data import DataLoader

test_ds = GeoSRDataset(f"{backup_root}/test", augment=False)
test_loader = DataLoader(test_ds, batch_size=4, shuffle=False, num_workers=0)

model.load_state_dict(torch.load(best_path, map_location="cpu", weights_only=True))
model.eval()

stats = {
    "model_sq": 0.0, "bicubic_sq": 0.0,
    "model_abs": 0.0, "bicubic_abs": 0.0,
    "count": 0, "sam_sum": 0.0, "sam_count": 0,
    "ndvi_sq": 0.0, "ndvi_count": 0,
    "pred_min": float("inf"), "pred_max": float("-inf")
}

with torch.no_grad():
    for lr, hr, mask in test_loader:
        lr, hr, mask = lr.to(device), hr.to(device), mask.to(device)

        with torch.amp.autocast("cuda", dtype=torch.float16):
            pred = model(lr).float()

        base = F.interpolate(
            lr, size=(256, 256), mode="bicubic", align_corners=False
        )

        m = mask.expand_as(hr)
        n = m.sum().item()

        stats["model_sq"] += (((pred-hr)**2)*m).sum().item()
        stats["bicubic_sq"] += (((base-hr)**2)*m).sum().item()
        stats["model_abs"] += ((pred-hr).abs()*m).sum().item()
        stats["bicubic_abs"] += ((base-hr).abs()*m).sum().item()
        stats["count"] += n

        dot = (pred*hr).sum(dim=1)
        norms = torch.linalg.vector_norm(pred, dim=1) * torch.linalg.vector_norm(hr, dim=1)
        valid = mask[:, 0] * (norms > 1e-8)
        angles = torch.acos((dot/(norms+1e-8)).clamp(-1,1))
        stats["sam_sum"] += (angles*valid).sum().item()
        stats["sam_count"] += valid.sum().item()

        ndvi_p = (pred[:,3]-pred[:,2])/(pred[:,3]+pred[:,2]+1e-8)
        ndvi_h = (hr[:,3]-hr[:,2])/(hr[:,3]+hr[:,2]+1e-8)
        nv = mask[:,0] * ((hr[:,3]+hr[:,2]).abs() > 1e-6)
        stats["ndvi_sq"] += (((ndvi_p-ndvi_h)**2)*nv).sum().item()
        stats["ndvi_count"] += nv.sum().item()

        stats["pred_min"] = min(stats["pred_min"], pred.min().item())
        stats["pred_max"] = max(stats["pred_max"], pred.max().item())

import math

count = stats["count"]
mse = stats["model_sq"]/count
bmse = stats["bicubic_sq"]/count

evaluation = {
    "test_patches": len(test_ds),
    "model_masked_MAE": stats["model_abs"]/count,
    "bicubic_masked_MAE": stats["bicubic_abs"]/count,
    "model_PSNR_dB_data_range_1": 10*math.log10(1/max(mse,1e-15)),
    "bicubic_PSNR_dB_data_range_1": 10*math.log10(1/max(bmse,1e-15)),
    "model_SAM_degrees": math.degrees(stats["sam_sum"]/max(stats["sam_count"],1)),
    "NDVI_RMSE": math.sqrt(stats["ndvi_sq"]/max(stats["ndvi_count"],1)),
    "prediction_min": stats["pred_min"],
    "prediction_max": stats["pred_max"],
    "ground_truth": "synthetic 10m HR patches",
    "test_split_used_for_training": False
}

with open(f"{output_dir}/evaluation_asia_test.json", "w") as f:
    json.dump(evaluation, f, indent=2)

for k, v in evaluation.items():
    print(k, ":", round(v, 6) if isinstance(v, float) else v)
```

```python
import os, torch
import torch.nn as nn
import torch.nn.functional as F

model_path = "/content/drive/MyDrive/GeoSR-AI/Final_Model_Asia_2026/GeoSRNet_Asia_x4_best.pth"

assert os.path.exists(model_path), "Asia checkpoint not found"

class ResBlock(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(48,48,3,padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(48,48,3,padding=1)
        )
    def forward(self,x):
        return x + 0.1*self.net(x)

class GeoSRNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.head = nn.Conv2d(4,48,3,padding=1)
        self.body = nn.Sequential(*[ResBlock() for _ in range(8)])
        self.up = nn.Sequential(
            nn.Conv2d(48,192,3,padding=1), nn.PixelShuffle(2), nn.ReLU(),
            nn.Conv2d(48,192,3,padding=1), nn.PixelShuffle(2), nn.ReLU()
        )
        self.tail = nn.Conv2d(48,4,3,padding=1)

    def forward(self,x):
        f = self.head(x)
        f = f + self.body(f)
        base = F.interpolate(x,scale_factor=4,mode="bicubic",
                             align_corners=False)
        return base + self.tail(self.up(f))

device = "cuda" if torch.cuda.is_available() else "cpu"
checkpoint = torch.load(model_path, map_location="cpu", weights_only=False)

if isinstance(checkpoint, dict):
    state = checkpoint.get("model_state_dict",
             checkpoint.get("state_dict", checkpoint))
else:
    state = checkpoint

model = GeoSRNet().to(device)
model.load_state_dict(state, strict=True)
model.eval()

print("Checkpoint loaded: PASS")
print("Parameters:", sum(p.numel() for p in model.parameters()))
print("Device:", device)
if device == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
import os, glob, time
import numpy as np
import rasterio
import torch
from rasterio.windows import Window
from rasterio.warp import reproject
from rasterio.enums import Resampling
from affine import Affine

folder = "/content/drive/MyDrive/GeoSR-AI/data/heldout_T42QWM"
out_path = "/content/drive/MyDrive/GeoSR-AI/GeoSR_AI_T42QWM_Asia_2p5m.tif"

assert not os.path.exists(out_path), "Output already exists; refusing to overwrite"

bands = ["B02", "B03", "B04", "B08"]
paths = {
    b: glob.glob(f"{folder}/*_{b}_10m.jp2")[0]
    for b in bands
}
scl_path = glob.glob(f"{folder}/*_SCL_20m.jp2")[0]

srcs = {b: rasterio.open(paths[b]) for b in bands}
scl_src = rasterio.open(scl_path)
ref = srcs["B02"]

H, W = ref.height, ref.width
assert all((s.width, s.height, s.transform, s.crs) ==
           (W, H, ref.transform, ref.crs) for s in srcs.values())

# Resample SCL to the 10 m grid using nearest neighbour.
scl10 = np.zeros((H, W), dtype=np.uint8)
reproject(
    source=rasterio.band(scl_src, 1),
    destination=scl10,
    src_transform=scl_src.transform,
    src_crs=scl_src.crs,
    dst_transform=ref.transform,
    dst_crs=ref.crs,
    resampling=Resampling.nearest
)

valid10 = ~np.isin(scl10, [0, 1, 3, 8, 9, 10])

profile = {
    "driver": "GTiff",
    "height": H * 4,
    "width": W * 4,
    "count": 4,
    "dtype": "float32",
    "crs": ref.crs,
    "transform": ref.transform * Affine.scale(0.25, 0.25),
    "nodata": -9999.0,
    "tiled": True,
    "blockxsize": 512,
    "blockysize": 512,
    "compress": "deflate",
    "predictor": 3,
    "BIGTIFF": "YES"
}

core, halo = 512, 32
device = "cuda"
model.eval()
start = time.time()
total = ((H + core - 1)//core) * ((W + core - 1)//core)
done = 0

with rasterio.open(out_path, "w", **profile) as dst:
    with torch.inference_mode():
        for y in range(0, H, core):
            for x in range(0, W, core):
                h = min(core, H-y)
                w = min(core, W-x)

                y0, x0 = max(0, y-halo), max(0, x-halo)
                y1, x1 = min(H, y+h+halo), min(W, x+w+halo)

                win = Window(x0, y0, x1-x0, y1-y0)
                arr = np.stack([
                    srcs[b].read(1, window=win).astype(np.float32)
                    for b in bands
                ])

                # Sentinel-2 BOA reflectance and model normalization.
                arr = (arr - 1000.0) / 10000.0 / 2.0
                inp = torch.from_numpy(arr).unsqueeze(0).to(device)

                pred = model(inp).float().cpu().numpy()[0] * 2.0

                cy, cx = (y-y0)*4, (x-x0)*4
                pred = pred[:, cy:cy+h*4, cx:cx+w*4]

                mask = np.repeat(
                    np.repeat(valid10[y:y+h, x:x+w], 4, axis=0),
                    4, axis=1
                )
                pred[:, ~mask] = -9999.0

                dst.write(
                    pred.astype(np.float32),
                    window=Window(x*4, y*4, w*4, h*4)
                )

                done += 1
                if done % 25 == 0 or done == total:
                    print(f"{done}/{total} tiles | "
                          f"{100*done/total:.1f}% | "
                          f"{(time.time()-start)/60:.1f} min")

for s in srcs.values():
    s.close()
scl_src.close()

print("\nINFERENCE COMPLETE")
print("Output:", out_path)
print("Size (GB):", round(os.path.getsize(out_path)/1024**3, 3))
```

```python
from pathlib import Path

api = Path("/content/drive/MyDrive/GeoSR-AI/inference_api")
api.mkdir(parents=True, exist_ok=True)

source = r'''
import os, shutil, tempfile, threading
from pathlib import Path
import numpy as np
import rasterio
from rasterio.windows import Window
from affine import Affine
import torch
import torch.nn as nn
import torch.nn.functional as F
from fastapi import FastAPI, UploadFile, File, HTTPException
from fastapi.responses import FileResponse
from starlette.background import BackgroundTask

MODEL_PATH = str(Path(__file__).parent.parent / "Final_Model_Asia_2026" / "GeoSRNet_Asia_x4_best.pth")
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
LOCK = threading.Lock()

class ResBlock(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(nn.Conv2d(48,48,3,padding=1),nn.ReLU(inplace=True),nn.Conv2d(48,48,3,padding=1))
    def forward(self,x):
        return x + 0.1*self.net(x)

class GeoSRNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.head = nn.Conv2d(4,48,3,padding=1)
        self.body = nn.Sequential(*[ResBlock() for _ in range(8)])
        self.up = nn.Sequential(nn.Conv2d(48,192,3,padding=1),nn.PixelShuffle(2),nn.ReLU(),nn.Conv2d(48,192,3,padding=1),nn.PixelShuffle(2),nn.ReLU())
        self.tail = nn.Conv2d(48,4,3,padding=1)
    def forward(self,x):
        f=self.head(x)
        f=f+self.body(f)
        return F.interpolate(x,scale_factor=4,mode="bicubic",align_corners=False)+self.tail(self.up(f))

model=GeoSRNet().to(DEVICE)
state=torch.load(MODEL_PATH,map_location="cpu",weights_only=True)
model.load_state_dict(state,strict=True)
model.eval()

app=FastAPI(title="GeoSR-AI Inference API",version="1.0")

@app.get("/health")
def health():
    return {"status":"ready","device":DEVICE,"parameters":sum(p.numel() for p in model.parameters()),"scale":4}

def infer(src_path,out_path):
    with rasterio.open(src_path) as src:
        if src.count != 4 or src.crs is None:
            raise ValueError("Input needs 4 bands and a valid CRS.")
        H,W=src.height,src.width
        profile=src.profile.copy()
        profile.update(driver="GTiff",height=H*4,width=W*4,count=4,dtype="float32",nodata=-9999.0,transform=src.transform*Affine.scale(.25,.25),compress="deflate",predictor=3,tiled=True,BIGTIFF="YES")
        with rasterio.open(out_path,"w",**profile) as dst, torch.inference_mode():
            for y in range(0,H,512):
                for x in range(0,W,512):
                    h,w=min(512,H-y),min(512,W-x)
                    y0,x0=max(0,y-32),max(0,x-32)
                    y1,x1=min(H,y+h+32),min(W,x+w+32)
                    raw=src.read(window=Window(x0,y0,x1-x0,y1-y0)).astype(np.float32)
                    valid=np.all(np.isfinite(raw),axis=0)
                    if src.nodata is not None:
                        valid &= np.all(raw!=src.nodata,axis=0)
                    raw[:,~valid]=0
                    inp=torch.from_numpy(raw/2).unsqueeze(0).to(DEVICE)
                    pred=model(inp).float().cpu().numpy()[0]*2
                    cy,cx=(y-y0)*4,(x-x0)*4
                    pred=pred[:,cy:cy+h*4,cx:cx+w*4]
                    mask=np.repeat(np.repeat(valid[y-y0:y-y0+h,x-x0:x-x0+w],4,axis=0),4,axis=1)
                    pred[:,~mask]=-9999
                    dst.write(pred.astype(np.float32),window=Window(x*4,y*4,w*4,h*4))

@app.post("/predict")
def predict(file:UploadFile=File(...)):
    if not file.filename.lower().endswith((".tif",".tiff")):
        raise HTTPException(400,"Upload a GeoTIFF.")
    folder=tempfile.mkdtemp()
    inp=os.path.join(folder,"input.tif")
    out=os.path.join(folder,"output_2p5m.tif")
    try:
        with open(inp,"wb") as f:
            shutil.copyfileobj(file.file,f)
        with LOCK:
            infer(inp,out)
        return FileResponse(out,media_type="image/tiff",filename="GeoSR_output_2p5m.tif",background=BackgroundTask(shutil.rmtree,folder,True))
    except Exception as e:
        shutil.rmtree(folder,ignore_errors=True)
        raise HTTPException(500,str(e))
'''

(api/"inference_api.py").write_text(source)
(api/"requirements.txt").write_text("fastapi\nuvicorn[standard]\npython-multipart\nnumpy\nrasterio\ntorch\n")
(api/"README.md").write_text(
    "# GeoSR-AI FastAPI\n\n"
    "Start: `uvicorn inference_api:app --host 0.0.0.0 --port 8000`\n\n"
    "Endpoints: GET /health, POST /predict.\n"
    "Input: four-band B02/B03/B04/B08 GeoTIFF in physical BOA reflectance.\n"
    "Normalization is performed internally (reflectance / 2).\n"
    "Output: four-band 4x GeoTIFF. A 2.5m output grid does not establish "
    "genuine 2.5m ground resolving capability.\n"
)

compile(source,"inference_api.py","exec")
print("API files created:")
for p in sorted(api.iterdir()):
    print(p.name,p.stat().st_size,"bytes")
print("Syntax check: PASS")
print("Checkpoint exists:",Path(source.split('MODEL_PATH = ')[1].splitlines()[0].strip('"').replace('Path(__file__).parent.parent / ','')) if False else (api.parent/"Final_Model_Asia_2026"/"GeoSRNet_Asia_x4_best.pth").exists())
```

```python
import glob, numpy as np, torch
import torch.nn.functional as F
from pathlib import Path
from collections import defaultdict

ROOT = Path("/content/drive/MyDrive/GeoSR-AI")
TEST_DIR = ROOT / "data/asia_dataset_v2/test"
MODEL_DIR = ROOT / "Final_Model_Asia_2026"

device = "cuda" if torch.cuda.is_available() else "cpu"
import sys
sys.path.insert(0, str(MODEL_DIR))
from geosrnet import GeoSRNet

model = GeoSRNet().to(device)
state = torch.load(MODEL_DIR/"GeoSRNet_Asia_x4_best.pth",
                   map_location="cpu", weights_only=True)
model.load_state_dict(state, strict=True)
model.eval()

files = sorted(glob.glob(str(TEST_DIR/"*.npz")))
bands = ["B02","B03","B04","B08"]
totals = {
    k: {"abs":np.zeros(4), "sq":np.zeros(4), "n":0}
    for k in ["model","bicubic"]
}
scenes = []

with torch.inference_mode():
    for file in files:
        z = np.load(file)
        hr, lr, masks = z["HR"], z["LR"], z["MASK"]

        s = {
            k: {"abs":0.0,"sq":0.0,"n":0}
            for k in ["model","bicubic"]
        }
        ndvi_sq = 0.0
        ndvi_n = 0

        for i in range(len(hr)):
            target = hr[i].astype(np.float32)       # 4,H,W
            low = lr[i].astype(np.float32)          # 4,h,w
            mask = masks[i,0].astype(bool)          # H,W

            x = torch.from_numpy(low).unsqueeze(0).to(device)
            pred = model(x).float()[0].cpu().numpy()

            bic = F.interpolate(
                x, size=target.shape[-2:],
                mode="bicubic", align_corners=False
            )[0].cpu().numpy()

            valid = mask & np.isfinite(target).all(axis=0)
            valid &= np.isfinite(pred).all(axis=0)

            if not valid.any():
                continue

            for name, output in [("model",pred),("bicubic",bic)]:
                err = output[:,valid] - target[:,valid]
                totals[name]["abs"] += np.abs(err).sum(axis=1)
                totals[name]["sq"] += (err**2).sum(axis=1)
                totals[name]["n"] += valid.sum()

                s[name]["abs"] += np.abs(err).sum()
                s[name]["sq"] += (err**2).sum()
                s[name]["n"] += err.size

            den_p = pred[3] + pred[2]
            den_t = target[3] + target[2]
            nv = valid & (np.abs(den_p)>1e-6) & (np.abs(den_t)>1e-6)

            if nv.any():
                pn = (pred[3,nv]-pred[2,nv])/den_p[nv]
                tn = (target[3,nv]-target[2,nv])/den_t[nv]
                ndvi_sq += np.square(pn-tn).sum()
                ndvi_n += nv.sum()

        model_mae = s["model"]["abs"]/s["model"]["n"]
        bic_mae = s["bicubic"]["abs"]/s["bicubic"]["n"]
        model_psnr = -10*np.log10(s["model"]["sq"]/s["model"]["n"])
        bic_psnr = -10*np.log10(s["bicubic"]["sq"]/s["bicubic"]["n"])

        scenes.append({
            "scene":Path(file).stem,
            "model_mae":model_mae,
            "bicubic_mae":bic_mae,
            "model_psnr":model_psnr,
            "bicubic_psnr":bic_psnr,
            "ndvi_rmse":np.sqrt(ndvi_sq/ndvi_n) if ndvi_n else np.nan
        })

print("SCENES:",len(scenes))
print("\nPER-BAND METRICS")
print("Band | Model MAE | Bicubic MAE | Model PSNR | Bicubic PSNR")
for j,b in enumerate(bands):
    n=totals["model"]["n"]
    mm=totals["model"]["abs"][j]/n
    bm=totals["bicubic"]["abs"][j]/n
    mp=-10*np.log10(totals["model"]["sq"][j]/n)
    bp=-10*np.log10(totals["bicubic"]["sq"][j]/n)
    print(f"{b} | {mm:.6f} | {bm:.6f} | {mp:.3f} | {bp:.3f}")

print("\nSCENE RESULTS")
for s in sorted(scenes,key=lambda x:x["model_mae"],reverse=True):
    print(
        f"{s['scene']}: "
        f"MAE {s['model_mae']:.6f} vs {s['bicubic_mae']:.6f} | "
        f"PSNR {s['model_psnr']:.2f} vs {s['bicubic_psnr']:.2f} | "
        f"NDVI RMSE {s['ndvi_rmse']:.5f}"
    )

wins=sum(s["model_mae"]<s["bicubic_mae"] for s in scenes)
print("\nMODEL WINS:",wins,"/",len(scenes))
print("MEDIAN MODEL MAE:",np.median([s["model_mae"] for s in scenes]))
print("WORST MODEL MAE:",max(s["model_mae"] for s in scenes))
```

```python
import glob, numpy as np, torch
import torch.nn.functional as F
from pathlib import Path
import sys

ROOT = Path("/content/drive/MyDrive/GeoSR-AI")
MD = ROOT/"Final_Model_Asia_2026"
sys.path.insert(0,str(MD))
from geosrnet import GeoSRNet

device = "cuda" if torch.cuda.is_available() else "cpu"
model = GeoSRNet().to(device)
model.load_state_dict(torch.load(
    MD/"GeoSRNet_Asia_x4_best.pth",
    map_location="cpu",weights_only=True
),strict=True)
model.eval()

thresholds = [0.0,0.02,0.05,0.10]
results = {t:{"model_sq":0.0,"bic_sq":0.0,"n":0,"total":0} for t in thresholds}
bad_scenes = []

with torch.inference_mode():
    for file in sorted(glob.glob(str(ROOT/"data/asia_dataset_v2/test/*.npz"))):
        z=np.load(file)
        scene={t:{"model_sq":0.0,"bic_sq":0.0,"n":0,"total":0} for t in thresholds}

        for i in range(len(z["HR"])):
            hr=z["HR"][i].astype(np.float32)
            lr=z["LR"][i].astype(np.float32)
            mask=z["MASK"][i,0].astype(bool)

            x=torch.from_numpy(lr).unsqueeze(0).to(device)
            pred=model(x).float()[0].cpu().numpy()
            bic=F.interpolate(x,size=hr.shape[-2:],mode="bicubic",
                              align_corners=False)[0].cpu().numpy()

            valid=mask & np.isfinite(hr).all(0) & np.isfinite(pred).all(0)
            den_t=hr[3]+hr[2]
            den_p=pred[3]+pred[2]
            den_b=bic[3]+bic[2]

            for t in thresholds:
                # Require nonnegative red/NIR and sufficient denominator
                v=valid & (hr[3]>=0) & (hr[2]>=0)
                v &= (pred[3]>=0) & (pred[2]>=0)
                v &= (den_t>t) & (den_p>t) & (den_b>t)

                scene[t]["total"] += valid.sum()
                results[t]["total"] += valid.sum()

                if not v.any():
                    continue

                target_ndvi=(hr[3,v]-hr[2,v])/den_t[v]
                model_ndvi=(pred[3,v]-pred[2,v])/den_p[v]
                bic_ndvi=(bic[3,v]-bic[2,v])/den_b[v]

                me=(model_ndvi-target_ndvi)**2
                be=(bic_ndvi-target_ndvi)**2

                scene[t]["model_sq"] += me.sum()
                scene[t]["bic_sq"] += be.sum()
                scene[t]["n"] += v.sum()

                results[t]["model_sq"] += me.sum()
                results[t]["bic_sq"] += be.sum()
                results[t]["n"] += v.sum()

        if scene[0.0]["n"]:
            raw=np.sqrt(scene[0.0]["model_sq"]/scene[0.0]["n"])
            strict=np.sqrt(scene[0.05]["model_sq"]/scene[0.05]["n"]) if scene[0.05]["n"] else np.nan
            if raw>0.2:
                bad_scenes.append((Path(file).stem,raw,strict))

print("GLOBAL NDVI AUDIT")
for t in thresholds:
    r=results[t]
    if r["n"]:
        print(f"Denominator > {t:.2f} | "
              f"Model RMSE={np.sqrt(r['model_sq']/r['n']):.5f} | "
              f"Bicubic RMSE={np.sqrt(r['bic_sq']/r['n']):.5f} | "
              f"Coverage={100*r['n']/r['total']:.2f}%")
    else:
        print("Threshold",t,"has no valid pixels")

print("\nSCENES WITH HIGH RAW NDVI ERROR")
for name,raw,strict in bad_scenes:
    print(name,"| raw:",round(raw,5),"| denominator >0.05:",round(strict,5))
```

```python
import glob, numpy as np, torch
import torch.nn.functional as F
from pathlib import Path
import sys

ROOT = Path("/content/drive/MyDrive/GeoSR-AI")
MD = ROOT / "Final_Model_Asia_2026"
sys.path.insert(0, str(MD))
from geosrnet import GeoSRNet

device = "cuda" if torch.cuda.is_available() else "cpu"
model = GeoSRNet().to(device)
model.load_state_dict(torch.load(
    MD / "GeoSRNet_Asia_x4_best.pth",
    map_location="cpu", weights_only=True
), strict=True)
model.eval()

scene_rows = []

with torch.inference_mode():
    for file in sorted(glob.glob(str(ROOT / "data/asia_dataset_v2/test/*.npz"))):
        z = np.load(file)
        model_abs = 0.0
        bic_abs = 0.0
        count = 0

        for i in range(len(z["HR"])):
            hr = z["HR"][i].astype(np.float32)
            lr = z["LR"][i].astype(np.float32)
            mask = z["MASK"][i, 0].astype(bool)

            x = torch.from_numpy(lr).unsqueeze(0).to(device)
            pred = model(x).float()[0].cpu().numpy()
            bic = F.interpolate(
                x, size=hr.shape[-2:], mode="bicubic",
                align_corners=False
            )[0].cpu().numpy()

            valid = mask & np.isfinite(hr).all(0)
            valid &= np.isfinite(pred).all(0) & np.isfinite(bic).all(0)

            model_abs += np.abs(pred - hr)[:, valid].sum()
            bic_abs += np.abs(bic - hr)[:, valid].sum()
            count += int(valid.sum()) * 4

        if count:
            scene_rows.append((
                Path(file).stem,
                model_abs / count,
                bic_abs / count
            ))

improvements = np.array([b - m for _, m, b in scene_rows])
relative = 100 * improvements / np.array([b for _, _, b in scene_rows])

rng = np.random.default_rng(42)
boot = np.empty(10000)

for i in range(len(boot)):
    sample = rng.choice(improvements, size=len(improvements), replace=True)
    boot[i] = sample.mean()

print("SCENE-LEVEL GENERALIZATION AUDIT")
print("Scenes:", len(scene_rows))
print("Model wins:", int((improvements > 0).sum()), "/", len(improvements))
print("Mean model MAE improvement:", round(improvements.mean(), 6))
print("Mean relative improvement (%):", round(relative.mean(), 2))
print("Median relative improvement (%):", round(np.median(relative), 2))
print("95% bootstrap CI for mean MAE improvement:",
      np.round(np.percentile(boot, [2.5, 97.5]), 6))
```

```python
import numpy as np, torch, glob
import torch.nn as nn
import torch.nn.functional as F
from pathlib import Path

ROOT = Path("/content/drive/MyDrive/GeoSR-AI")
MD = ROOT / "Final_Model_Asia_2026"

class ResBlock(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(48,48,3,padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(48,48,3,padding=1)
        )
    def forward(self,x):
        return x + 0.1*self.net(x)

class GeoSRNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.head = nn.Conv2d(4,48,3,padding=1)
        self.body = nn.Sequential(*[ResBlock() for _ in range(8)])
        self.up = nn.Sequential(
            nn.Conv2d(48,192,3,padding=1), nn.PixelShuffle(2), nn.ReLU(),
            nn.Conv2d(48,192,3,padding=1), nn.PixelShuffle(2), nn.ReLU()
        )
        self.tail = nn.Conv2d(48,4,3,padding=1)

    def forward(self,x):
        f = self.head(x)
        f = f + self.body(f)
        return F.interpolate(
            x,scale_factor=4,mode="bicubic",align_corners=False
        ) + self.tail(self.up(f))

device = "cuda" if torch.cuda.is_available() else "cpu"
model = GeoSRNet().to(device)

state = torch.load(
    MD/"GeoSRNet_Asia_x4_best.pth",
    map_location="cpu", weights_only=True
)
model.load_state_dict(state, strict=True)
model.eval()

groups = {
    "Low NDVI (<0.20)": (-1.0, 0.20),
    "Moderate NDVI (0.20-0.50)": (0.20, 0.50),
    "High NDVI (>0.50)": (0.50, 1.01)
}
stats = {k:[0.0,0.0,0] for k in groups}

with torch.inference_mode():
    files = sorted(glob.glob(
        str(ROOT/"data/asia_dataset_v2/test/*.npz")
    ))

    for f in files:
        z = np.load(f)
        for i in range(len(z["HR"])):
            hr = z["HR"][i]
            mask = z["MASK"][i,0].astype(bool)
            x = torch.from_numpy(z["LR"][i:i+1]).to(device)

            pred = model(x).float()[0].cpu().numpy()
            bic = F.interpolate(
                x,size=hr.shape[-2:],mode="bicubic",
                align_corners=False
            )[0].cpu().numpy()

            den = hr[3]+hr[2]
            valid = mask & (hr[3]>=0) & (hr[2]>=0) & (den>0.05)

            ndvi = np.zeros_like(den)
            ndvi[valid] = (hr[3,valid]-hr[2,valid])/den[valid]

            me = np.abs(pred-hr).mean(axis=0)
            be = np.abs(bic-hr).mean(axis=0)

            for name,(lo,hi) in groups.items():
                v = valid & (ndvi>=lo) & (ndvi<hi)
                if v.any():
                    stats[name][0] += me[v].sum()
                    stats[name][1] += be[v].sum()
                    stats[name][2] += int(v.sum())

print("NDVI-STRATIFIED ROBUSTNESS AUDIT")
for name,(ms,bs,n) in stats.items():
    if n:
        print(
            f"{name} | Pixels={n:,} | "
            f"Model MAE={ms/n:.6f} | Bicubic MAE={bs/n:.6f} | "
            f"Improvement={100*(bs-ms)/bs:.2f}%"
        )
    else:
        print(name,"| No valid pixels")
```
